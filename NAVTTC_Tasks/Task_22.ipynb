```python
# empty tuple
my_tuple = ()

# tuple having integers
my_tuple = (1, 2, 3)

# tuple with mixed datatypes
my_tuple = (1, "Hello", 3.4)

# nested tuple
my_tuple = ("mouse", [8, 4, 6], (1, 2, 3))

# tuple can be created without parentheses
# also called tuple packing
my_tuple = 3, 4.6, "dog"
# tuple unpacking is also possible
a, b, c = my_tuple
```

```python
my_tuple = ("hello")   # only parentheses is not enough
print(type(my_tuple))

my_tuple = ("hello",)  # need a comma at the end
print(type(my_tuple))

my_tuple = "hello",    # parentheses is optional
print(type(my_tuple))
```

```python
my_tuple = ['p', 'e', 'r', 'm', 'i', 't']

print(my_tuple[0])
print(my_tuple[5])

# Index must be in range
# print(my_tuple[6])  # This gives IndexError

# Index must be an integer
# print(my_tuple[2.0])  # This gives TypeError


n_tuple = ("mouse", [8, 4, 6], (1, 2, 3))

# Nested indexing
print(n_tuple[0][3])
print(n_tuple[1][1])
print(n_tuple[2][0])
```

```python
my_tuple = ['p', 'e', 'r', 'm', 'i', 't']

print(my_tuple[-1])
print(my_tuple[-6])
```

```python
my_tuple = ('p', 'r', 'o', 'g', 'r', 'a', 'm', 'i', 'z')

print(my_tuple[1:4])   # elements 2nd to 4th
print(my_tuple[:-7])  # elements beginning to 2nd
print(my_tuple[7:])   # elements 8th to end
print(my_tuple[:])    # elements beginning to end
```

```python
my_tuple = (4, 2, 3, [6, 5])

# We cannot change a tuple element
# my_tuple[1] = 9
# This gives: TypeError

# We also cannot replace the list itself
# my_tuple[3] = 9
# This gives: TypeError

# But we CAN change an element inside the list
my_tuple[3][0] = 9

print(my_tuple)

# Tuples themselves can be reassigned
my_tuple = ('p', 'r', 'o', 'g', 'r', 'a', 'm', 'i', 'z')

print(my_tuple)
```

```python
# Tuple concatenation
result1 = (1, 2, 3) + (4, 5, 6)
print(result1)

# Tuple repetition
result2 = ("Repeat",) * 3
print(result2)
```

```python
my_tuple = ('p', 'r', 'o', 'g', 'r', 'a', 'm', 'i', 'z')

# Cannot delete an individual tuple item
# del my_tuple[3]
# This gives: TypeError

# Can delete the entire tuple
del my_tuple

# Now my_tuple no longer exists
# print(my_tuple)
# This gives: NameError
```

```python
my_tuple = ('a', 'p', 'p', 'l', 'e')

print(my_tuple.count('p'))
print(my_tuple.index('l'))
```

```python
my_tuple = ('a', 'p', 'p', 'l', 'e')

print('a' in my_tuple)
print('b' in my_tuple)
print('g' not in my_tuple)
```
